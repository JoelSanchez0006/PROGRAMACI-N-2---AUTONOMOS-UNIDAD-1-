# **Herencia y Sintaxis de Herencia - Métodos Mágicos y Sobrecarga de Operadores**
a) Crea una clase base ModeloIA con metodos
 - entrenar
 - predecir
 - evaluar

a) Implementación en la clase base ModeloIA modelos especiales
- __ str __
- __ add __
- __ sub __

b) Se crea una clase donde se sumen y resten modelos y se imprimen sus resultados

```python
# Clase base con nombre modeloia y define la estructura y el comportamiento comun para todos los modelos
class modeloia:
  # Inicializa el nombre y la precisión del modelo.
  def __init__(self, nombre="MODELO001", precision=0.0):
    self.nombre = nombre
    self.precision = precision
    print(f"ModeloIA: '{self.nombre}' inicializado con precisión {self.precision:.2f}.")

  def entrenar(self, datos):
    print(f"Modelo '{self.nombre}' entrenando con: {datos}")

  def predecir(self, entrada):
    print(f"Modelo '{self.nombre}' prediciendo para: {entrada}")
    return f"Predicción de {self.nombre} para {entrada}"

  def evaluar(self, resultados_reales, resultados_predichos):
    print(f"Modelo '{self.nombre}' evaluando: reales={resultados_reales} y predichos={resultados_predichos}")

  # Método mágico __str__ para una representación legible del objeto.
  def __str__(self):
    return f"ModeloIA: {self.nombre} (Precisión: {self.precision:.2f})"

  # Método mágico __add__ para definir el comportamiento del operador '+'.
  def __add__(self, other):
    if not isinstance(other, modeloia):
      raise TypeError("Solo se pueden sumar objetos de tipo ModeloIA")
    new_nombre = f"{self.nombre} + {other.nombre}"
    # Calcula la precisión promedio.
    new_precision = (self.precision + other.precision) / 2
    print(f"Fusionando modelos: {self.nombre} y {other.nombre}")
    return modeloia(nombre=new_nombre, precision=new_precision)

  # Método mágico __sub__ para definir el comportamiento del operador '-'.
  def __sub__(self, other):
    if not isinstance(other, modeloia):
      raise TypeError("Solo se pueden restar objetos de tipo ModeloIA")
    new_nombre = f"{self.nombre} - {other.nombre}"
    # Calcula la diferencia absoluta de precisión.
    new_precision = abs(self.precision - other.precision)
    print(f"Calculando diferencia entre modelos: {self.nombre} y {other.nombre}")
    return modeloia(nombre=new_nombre, precision=new_precision)
```

# **CREACION DE CLASES HIJAS Y FUNCION SUPER**
 - Clasificador IA
 - Regresor IA

```python
# Clases hijas que heredan de modeloia.
class ClasificadorIA(modeloia):
  def __init__(self, nombre="ClasificadorIA", precision=0.0, tipo_clasificador="General"):
    super().__init__(nombre, precision)
    self.tipo_clasificador = tipo_clasificador
    print(f"ClasificadorIA '{self.nombre}' (Tipo: {self.tipo_clasificador}) inicializado.")

  # Sobreescribe el método entrenar de la clase padre.
  def entrenar(self, datos):
    # Llama al método entrenar de la clase padre.
    super().entrenar(datos)
    print(f"ClasificadorIA '{self.nombre}': Entrenando para clasificar con {datos}.")

  # Sobreescribe el método predecir de la clase padre para una predicción específica de clasificación.
  def predecir(self,entrada):
    # Llama al método predecir de la clase padre.
    super().predecir(entrada)
    print(f"ClasificadorIA '{self.nombre}': Prediciendo categoría para {entrada}.")
    return f"ClasificadorIA: Predicción para {entrada}"


# Clase para modelos de IA de regresión.
class RegresorIA(modeloia):
  # Llama al constructor de la clase padre y añade un atributo para el tipo de regresor.
  def __init__(self, nombre="RegresorIA", precision=0.0, tipo_regresor="Lineal"):
    super().__init__(nombre, precision)
    self.tipo_regresor = tipo_regresor
    print(f"RegresorIA '{self.nombre}' (Tipo: {self.tipo_regresor}) inicializado.")

  # Sobreescribe el método entrenar de la clase padre.
  def entrenar(self, datos):
    super().entrenar(datos)
    print(f"RegresorIA '{self.nombre}': Entrenando para regresión con {datos}.")

  # Sobreescribe el método predecir de la clase padre para una predicción específica de regresión.
  def predecir(self, entrada):
    super().predecir(entrada)
    print(f"RegresorIA '{self.nombre}': Prediciendo valor numérico para {entrada}.")
    return f"RegresorIA: Predicción para {entrada}"
```

# **Polimorfismo**

```python
# Crear objetos
modelo = modeloia(nombre="Modelo Base A", precision=0.75)
clasificador = ClasificadorIA(nombre="MiClasificador", precision=0.95, tipo_clasificador="Multiclase")
regresor = RegresorIA(nombre="MiRegresor", precision=0.88, tipo_regresor="Polinomial")
print("\n")
# Ejecutar entrenamientos
clasificador.entrenar("datos de clasificacion")
regresor.entrenar("datos numéricos")

# Lista con objetos de diferentes tipos
modelos = [
    modelo,
    clasificador,
    regresor
]

# Polimorfismo
print(" \n ")
print("=== Demostración de Polimorfismo ===")
for modelo_item in modelos:
  print("\n")
  print(f"Predicción de {modelo_item.nombre}: {modelo_item.predecir('los datos ingresados')}")
```

# **Patron singleton y Gestion de sesiones de entrenamiento**

```python
# Implementación del Patrón Singleton para la gestión de sesiones.
class GestorSesion:
  # Atributo de clase para almacenar la única instancia.
    _instance = None

    # Método mágico __new__ se llama antes que __init__ para controlar la creación de la instancia.
    def __new__(cls):
        # Si no hay una instancia existente, crea una nueva.
        if cls._instance is None:
            cls._instance = super(GestorSesion, cls).__new__(cls)
            cls._instance.sesion_activa = False
            print("Gestor de Sesión creado.")
        else:
            # Si ya existe una instancia, reutiliza la existente.
            print("Reutilizando Gestor de Sesión.")
        return cls._instance

    # Método para iniciar una sesión.
    def iniciar_sesion(self):
        # Solo inicia la sesión si no está activa actualmente.
        if not self.sesion_activa:
            self.sesion_activa = True
            print("Sesión iniciada.")
        else:
            print("Sesión ya activa.")

    # Método para cerrar una sesión.
    def cerrar_sesion(self):
        # Solo cierra la sesión si está activa actualmente.
        if self.sesion_activa:
            self.sesion_activa = False
            print("Sesión cerrada.")
        else:
            print("No hay sesión activa.")

```

# **Probar la funcionalidad de iniciar y cerrar sesión.**




```python
# Demostración del Patrón Singleton
print("--- Demostración de Singleton ---")

# La primera vez se crea la instancia.
gestor1 = GestorSesion()
# La segunda vez se reutiliza la instancia existente.
gestor2 = GestorSesion()

# Comprobar que ambas variables referencian el mismo objeto.
print(f"ID gestor1: {id(gestor1)}")
print(f"ID gestor2: {id(gestor2)}")
print(f"¿Son el mismo objeto? {gestor1 is gestor2}")

# Probar la funcionalidad de iniciar y cerrar sesión.
gestor1.iniciar_sesion() # Inicia la sesión.
gestor2.iniciar_sesion() # Intenta iniciar la sesión de nuevo, pero ya está activa.

gestor1.cerrar_sesion() # Cierra la sesión.
gestor2.cerrar_sesion() # Intenta cerrar la sesión de nuevo, pero ya está cerrada.
gestor1.cerrar_sesion() # Intenta cerrar la sesión de nuevo, pero ya está cerrada.
```
