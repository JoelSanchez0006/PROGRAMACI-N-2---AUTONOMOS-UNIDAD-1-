**1. Creación del arreglo NumPy**

Crear un arreglo bidimensional llamado consumo con los datos de la
tabla.
Luego mostrar:
1. La matriz completa.
2. El número de dimensiones.
3. La forma del arreglo.
4. El número total de elementos.
5. El tipo de dato almacenado.



```python
# Importar la librería NumPy
import numpy as np

# Crear el arreglo bidimensional llamado consumo
consumo = np.array([
    [12.5, 13.0, 12.8, 14.1, 13.7, 6.2, 4.8],   #Computadoras
    [6.5, 6.8, 7.0, 6.9, 7.2, 3.1, 2.5],        #Laptops
    [18.0, 18.2, 18.1, 18.3, 18.4, 18.0, 17.9], #Servidor local
    [4.2, 4.1, 4.3, 4.4, 4.2, 4.0, 3.9],        # Red
    [9.0, 9.5, 9.2, 9.8, 10.0, 3.5, 2.0]        #Iluminacion
])

# Mostrar la matriz completa
print("Matriz completa:")
print(consumo)

# Mostrar el número de dimensiones
print("\nNumero de dimensiones:")
print(consumo.ndim)

# Mostrar la forma del arreglo
print("\nForma del arreglo:")
print(consumo.shape)

# Mostrar el número total de elementos
print("\nNumero total de elementos:")
print(consumo.size)

# Mostrar el tipo de dato almacenado
print("\nTipo de dato almacenado:")
print(consumo.dtype)
```

**2. Análisis general del consumo**

Usando funciones de NumPy, calcular:
1. Consumo total semanal de todos los dispositivos.
2. Consumo promedio general.
3. Consumo máximo registrado.
4. Consumo mínimo registrado.
Después, escribir una interpretación breve sobre el comportamiento
general del consumo energético.

```python
#Consumo total semanal
total_consumo = np.sum(consumo)

#Promedio general
promedio_general = np.mean(consumo)

#Consumo maximo
consumo_maximo = np.max(consumo)

#Consumo minimo
consumo_minimo = np.min(consumo)

#Para poder tener solo dos decimales
resultado = round(promedio_general, 2)

print("Consumo total semanal:", total_consumo)
print("Consumo promedio general:", resultado)
print("Consumo máximo registrado:", consumo_maximo)
print("Consumo mínimo registrado:", consumo_minimo)
```

**¿Qué conclusión se puede obtener sobre el servidor local?**

**6. Mini análisis con condición**

Crear una condición para identificar los consumos mayores a 15 kWh.
Luego responder:
1. ¿Cuántos registros superan los 15 kWh?
2. ¿A qué dispositivo pertenecen principalmente esos valores?
3. ¿Qué indica esto sobre el patrón de consumo?

```python

# Crear condición para consumos mayores a 15
mayores_15 = consumo > 15

# Obtener valores mayores a 15
valores_mayores = consumo[mayores_15]

# Cantidad de registros
print("\nCantidad de registros mayores a 15:")
print(valores_mayores.size)

# Mostrar qué dispositivo tiene esos consumos
print("\nLos dispositivos que con consumos mayores a 15 kWh pertenecen a: ")

for i in range(consumo.shape[0]):

    if np.any(consumo[i] > 15):
        print(dispositivos[i])
```

**¿Cuántos registros superan los 15 kWh?**

Se registran 7 registros que superan los 15

**¿A qué dispositivo pertenecen principalmente esos valores?**
Todos los dispositivos pertecen al servidor local

**¿Qué indica esto sobre el patrón de consumo?**

Con lo que se puede identificar es que el servicio local matiene un consumo muy alto respecto a lo que es toda la semana
