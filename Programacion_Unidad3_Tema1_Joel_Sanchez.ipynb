# **Detección y tratamiento de valores atípicos (percentiles, z-score, IQR)**

```python
import matplotlib.pyplot as plt
from scipy.stats import zscore
```

Boxplots originales

```python
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.boxplot(df['Unit price'])
plt.title('Unit price antes')

plt.subplot(1,2,2)
plt.boxplot(df['Quantity'])
plt.title('Quantity antes')

plt.show()
```

# **• Detecta outliers usando IQR.**

IQR

```python
Q1 = df['Unit price'].quantile(0.25)
Q3 = df['Unit price'].quantile(0.75)

IQR = Q3 - Q1

limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

outliers_iqr = df[(df['Unit price'] < limite_inferior) |
                  (df['Unit price'] > limite_superior)]

print("Cantidad de outliers con IQR:", len(outliers_iqr))
```

columna Quantity

```python
Q1_q = df['Quantity'].quantile(0.25)
Q3_q = df['Quantity'].quantile(0.75)

IQR_q = Q3_q - Q1_q

lim_inf_q = Q1_q - 1.5*IQR_q
lim_sup_q = Q3_q + 1.5*IQR_q

outliers_quantity = df[(df['Quantity'] < lim_inf_q) |
                       (df['Quantity'] > lim_sup_q)]

print("Cantidad de outliers:", len(outliers_quantity))
```

# **Detecta outliers usando z-score**

```python
z_unitprice = zscore(df['Unit price'])

outliers_z = df[abs(z_unitprice) > 3]

print("Outliers por z-score:", len(outliers_z))
```

```python
z_quantity = zscore(df['Quantity'])

outliers_z_quantity = df[abs(z_quantity) > 3]

print("Outliers por z-score:", len(outliers_z_quantity))
```

# **• Aplica un tratamiento**

```python
df_outliers = df.copy()

df_outliers['Unit price'] = np.where(
    df_outliers['Unit price'] > limite_superior,
    limite_superior,
    np.where(
        df_outliers['Unit price'] < limite_inferior,
        limite_inferior,
        df_outliers['Unit price']
    )
)
```

```python
df_outliers['Quantity'] = np.where(
    df_outliers['Quantity'] > lim_sup_q,
    lim_sup_q,
    np.where(
        df_outliers['Quantity'] < lim_inf_q,
        lim_inf_q,
        df_outliers['Quantity']
    )
)
```

# **Creación de nuevas variables (feature engineering)**

```python
# Variable 1
df_outliers['is_high_value'] = df_outliers['Sales'] > df_outliers['Sales'].mean()
# Variable 2
df_outliers['price_category'] = pd.cut(
    df_outliers['Unit price'],
    bins=3,
    labels=['Bajo', 'Medio', 'Alto']
)
# Variable 3
df_outliers['sales_per_unit'] = df_outliers['Sales'] / df_outliers['Quantity']
# Variable 4
df_outliers['is_weekend'] = df_outliers['Weekend'].astype(int)
promedio_branch = df_outliers.groupby('Branch')['Sales'].mean()
# Variable 5
df_outliers['avg_branch_sales'] = df_outliers['Branch'].map(promedio_branch)
```

```python
df_outliers[['is_high_value',
             'price_category',
             'sales_per_unit',
             'is_weekend',
             'avg_branch_sales']].head()
```

| Nueva variable   | Descripción                                        | Fórmula                         |
| ---------------- | -------------------------------------------------- | ------------------------------- |
| is_high_value    | Indica si una venta es superior al promedio        | Total > promedio(Total)         |
| price_category   | Clasifica el precio unitario                       | pd.cut(Unit price)              |
| sales_per_unit   | Venta promedio por unidad                          | Total / Quantity                |
| is_weekend       | Indica si la compra fue realizada en fin de semana | Weekend → 0 o 1                 |
| avg_branch_sales | Promedio de ventas por sucursal                    | groupby(Branch)['Total'].mean() |
